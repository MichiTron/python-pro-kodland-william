# Meme con pie de foto (115656)
Añadimos un texto blanco centrado abajo de una imagen con OpenCV.

## 1. Subir la imagen
Sube una foto (por ejemplo, tu meme con el fondo cambiado).

```python
from google.colab import files

subida = files.upload()
ruta_imagen = list(subida.keys())[0]
```

## 2. Función que agrega el pie de foto
`cv2.getTextSize` da el ancho y alto del texto; con eso calculamos `x` para centrarlo y `y` para dejarlo abajo.

```python
import cv2

from google.colab.patches import cv2_imshow


def agregar_pie_de_foto(ruta_imagen, texto, ruta_salida):
    image = cv2.imread(ruta_imagen)
    if image is None:
        raise FileNotFoundError(f"No se pudo cargar la imagen: {ruta_imagen}")

    font = cv2.FONT_HERSHEY_SIMPLEX
    font_scale = 1
    color = (255, 255, 255)  # blanco (B, G, R)
    thickness = 2

    # Tamano del texto para centrarlo horizontalmente y dejarlo abajo con un margen
    (ancho_texto, alto_texto), base = cv2.getTextSize(texto, font, font_scale, thickness)
    alto, ancho = image.shape[:2]
    x = (ancho - ancho_texto) // 2
    y = alto - base - 20

    image_with_text = cv2.putText(image, texto, (x, y), font, font_scale, color, thickness, cv2.LINE_AA)
    cv2_imshow(image_with_text)
    cv2.imwrite(ruta_salida, image_with_text)
    return image_with_text
```

## 3. Probar
El resultado se muestra y se guarda como `meme_con_pie.jpg`.

```python
agregar_pie_de_foto(ruta_imagen, "Meme realizado por mí", "meme_con_pie.jpg")
```
